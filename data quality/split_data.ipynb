```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("data/data_limpa.csv")

print("Shape original:", df.shape)
```

```python
target = "TriageGrade"
```

```python
train_df, temp_df = train_test_split(
    df,
    test_size=0.35,          # 15% + 20%
    random_state=42,
    stratify=df[target]
)
```

```python
val_df, test_df = train_test_split(
    temp_df,
    test_size=20/35,
    random_state=42,
    stratify=temp_df[target]
)
```

```python
print("Train:", train_df.shape)
print("Validation:", val_df.shape)
print("Test:", test_df.shape)

print("\nDistribuição da target:")
print("Train:\n", train_df[target].value_counts(normalize=True))
print("Validation:\n", val_df[target].value_counts(normalize=True))
print("Test:\n", test_df[target].value_counts(normalize=True))
```

```python
train_df.to_csv("data/train.csv", index=False)
val_df.to_csv("data/val.csv", index=False)
test_df.to_csv("data/test.csv", index=False)
```
